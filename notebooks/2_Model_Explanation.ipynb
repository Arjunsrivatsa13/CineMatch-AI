# Movie Recommendation System - Model Explanation

## Overview

This project uses a content-based recommendation system.
It recommends movies based on similarity between movie genres.

The system uses:
- TF-IDF Vectorization
- Cosine Similarity

## 1. TF-IDF Vectorization

TF-IDF (Term Frequency - Inverse Document Frequency) converts text data into numerical vectors.

In this project, movie genres are converted into vectors so that movies can be compared mathematically.

- Each movie is represented by its genre information.
- Similar genres create similar vector representations.

## 2. Cosine Similarity

Cosine Similarity measures how similar two movie vectors are.

The similarity score is calculated by comparing the angle between two vectors.

- A score closer to 1 means movies are more similar.
- A score closer to 0 means movies are less similar.

The system uses these similarity scores to find movies that are most similar to the selected movie.


## 3. Recommendation Process

The recommendation system follows these steps:

1. Load the movie dataset.
2. Extract movie genre information.
3. Convert genres into TF-IDF vectors.
4. Calculate cosine similarity between movies.
5. Find movies with the highest similarity scores.
6. Display the top recommended movies.

## Conclusion

This content-based recommendation system successfully recommends movies by analyzing genre similarity.

TF-IDF and cosine similarity help identify movies with similar characteristics and provide personalized recommendations.
