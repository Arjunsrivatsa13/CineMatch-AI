```python
import pandas as pd
import matplotlib.pyplot as plt

movies = pd.read_csv("../data/movies.csv")

movies.head()
```

```python
movies.info()
```

```python
genre_counts = movies["genres"].str.split("|").explode().value_counts()

genre_counts.head(10).plot(
    kind="bar",
    figsize=(10,5),
    title="Top 10 Movie Genres"
)

plt.xlabel("Genre")
plt.ylabel("Number of Movies")
plt.show()
```

```python
movies["year"] = movies["title"].str.extract(r"\((\d{4})\)")

movies["year"].value_counts().sort_index().plot(
    figsize=(12,5),
    title="Movies Released Over Years"
)

plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.show()
```

```python
ratings = pd.read_csv("../data/ratings.csv")

ratings.head()
```

```python
avg_rating = ratings.groupby("movieId")["rating"].mean()

avg_rating.head()
```

```python
ratings["rating"].value_counts().sort_index().plot(
    kind="bar",
    figsize=(8,4),
    title="Rating Distribution"
)

plt.xlabel("Rating")
plt.ylabel("Number of Ratings")
plt.show()
```

## Conclusion

- The dataset contains 9,742 movies with genre information.
- Genre analysis shows the most common movie categories.
- Movie release trends help understand changes in movie production over time.
- Rating analysis shows user rating patterns and preferences.
- The dataset is clean and suitable for building a movie recommendation system.
